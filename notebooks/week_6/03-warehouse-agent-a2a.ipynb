```python
import httpx
from a2a.client import A2ACardResolver, A2AClient, ClientFactory
from a2a.types import (
    AgentCard,
    Message,
    MessageSendParams,
    Part,
    Role,
    SendMessageRequest,
    TextPart,
)
```

```python
BASE_URL="http://localhost:8020"
PUBLIC_AGENT_CARD_PATH ="/.well-known/agent.json"
```

```python
import pprint


async with httpx.AsyncClient() as client:
    
    resolver = A2ACardResolver(
        httpx_client=client,
        base_url=BASE_URL,
    )
    
    card = await resolver.get_agent_card(PUBLIC_AGENT_CARD_PATH)
    
    pprint.pprint(card)
```

```python
import uuid

timeoout_config = httpx.Timeout(connect=10.0, read=100.0, write=10.0, pool=10.0)

client = await ClientFactory.connect(BASE_URL, resolver_http_kwargs={"timeout": timeoout_config})

message_id = str(uuid.uuid4())

message = Message(
    message_id=message_id,
    metadata={"user_id": "krishnak"},
    context_id="thread-0001",
    role=Role.user,
    parts=[Part(root=TextPart(text="Hello, how'z weather in Bangalore?"))],
)

final_response = None

async for response in client.send_message(message):
    final_response = response

for artifact in final_response[0].artifacts:
    for part in artifact.parts:
        print(part.root.text)
```
